# NumPy (넘파이)

다차원 배열을 다룰 수 있는 기능을 제공하는 라이브러리

- 넘파이 배열 : **ndarray**

```python
import numpy as np
```

## 1. 파이썬 리스트 vs 넘파이 배열

파이썬 리스트에는 **다양한 타입의 데이터**를 담을 수 있음.

```python
data_list = [1, "이", 3.0, [4]]
print(f"data_list: {data_list}")
print(f"각 요소 별 타입 : {[type(x).__name__ for x in data_list]}")
```

리스트는 데이터를 구분하지 않고 담을 수 있는 점이 편리할 수 있지만,
**다른 리스트와 연산을 하고자 할 때는 문제가 될 수 있음!**

(아래 셀을 실행하면 `"이" + 20`에서 **TypeError**가 나는 것을 확인할 수 있음)

```python
data_list2 = [10, 20, 30, 40]

print("-" * 60)
for d1, d2 in zip(data_list, data_list2):
    print(f"{d1} + {d2} = {d1 + d2}")
print("-" * 60)
```

넘파이 배열(ndarray)은 **데이터 타입이 하나로 고정된 구조** → 연산이 쉬움!

```python
arr = np.array([1, 2, 3, 4])
print(f"arr : {arr}")
print(f"배열 타입 : {arr.dtype}")

arr2 = np.array([5, 6, 7, 8])
print(f"arr2 : {arr2}")
```

반복문으로 하나씩 더하기 (리스트 방식)

```python
for d1, d2 in zip(arr, arr2):
    print(f"{d1} + {d2} = {d1 + d2}")
```

배열 단위로 한 번에 더하기 (**벡터화 연산**)

넘파이 배열은 자바의 배열과 비슷하지만, 연산 수행 시 **반복문을 사용하지 않고** 배열 단위의 연산이 가능

```python
print(f"arr + arr2 = {arr + arr2}")
```

## 2. 넘파이 배열의 차원, 형태

| 차원 | 이름 | 설명 |
|---|---|---|
| 0차원 | 스칼라 | 단일 값 |
| 1차원 | 벡터 | 0차원의 모임 |
| 2차원 | 행렬 | 1차원의 모임 |

```python
# 0차원 배열 -> 스칼라 (단일 값)
arr_0d = np.array(42)
print(f"0차원 배열 : {arr_0d}, 차원: {arr_0d.ndim}, 형태: {arr_0d.shape}")
```

```python
# 1차원 배열 -> 벡터 (0차원의 모임)
arr_1d = np.array([1, 2, 3])
print(f"1차원 배열 : {arr_1d}, 차원: {arr_1d.ndim}, 형태: {arr_1d.shape}")
```

```python
# 2차원 배열 -> 행렬 (1차원의 모임)
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print(f"2차원 배열 :\n{arr_2d}")
print(f"차원: {arr_2d.ndim}, 형태: {arr_2d.shape}")
```

## 3. 리스트 vs 배열 메모리 크기 비교

숫자 표현 시 `_`로 구분 가능 (`100_000` = `100000`)

```python
n = 100_000
print(f"n: {n:,}")

# 리스트랑 배열을 0부터 99_999 데이터로 생성
list_10m = list(range(n))
arr_10m = np.arange(n)

print(f"list_10m : {list_10m[:5]}...{list_10m[-5:]}")
print(f"arr_10m  : {arr_10m[:5]}...{arr_10m[-5:]}")
```

```python
import sys

# 리스트: 리스트 자체(칸 목록) + 정수 객체 하나하나의 크기
list_container = sys.getsizeof(list_10m)
int_obj = sys.getsizeof(list_10m[0])
list_bytes = list_container + (int_obj * n)

# 배열: 실제 데이터가 차지하는 바이트
arr_bytes = arr_10m.nbytes

print(f"데이터 {n:,}개 기준")
print(f" - 리스트: {list_bytes / 1024 / 1024:6.2f}MB")
print(f" - 배열  : {arr_bytes / 1024 / 1024:6.2f}MB")
```

→ 10만 개 기준으로 약 2~3MB 정도 차이

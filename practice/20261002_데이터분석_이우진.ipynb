### 4. Age, Fare의 평균·최솟값·최댓값

```python
import pandas as pd
df = pd.read_csv("train.csv")

print("Age 평균:", df["Age"].mean())
print("Age 최댓값:", df["Age"].max())
print("Age 최솟값:", df["Age"].min())

print("Fare 평균:", df["Fare"].mean())
print("Fare 최댓값:", df["Fare"].max())
print("Fare 최솟값:", df["Fare"].min())
```

### 5. 생존자·사망자 수

```python
import pandas as pd
df =pd.read_csv("train.csv")

counts = df["Survived"].value_counts()

print(f"사망자(0): {counts[0]}명")
print(f"생존자(1): {counts[1]}명")
```

### 8. 나이대(AgeGroup) 열 추가

```python
def age_group(age):
    if pd.isna(age):
        return "미확인"
    elif age < 10:
        return "아동"
    elif age < 20:
        return "10대"
    elif age < 30:
        return "20대"
    elif age < 40:
        return "30대"
    elif age < 50:
        return "40대"
    elif age < 60:
        return "50대"
    else:
        return "60대 이상"

df["AgeGroup"]= df["Age"].apply(age_group)    

df.head()
```

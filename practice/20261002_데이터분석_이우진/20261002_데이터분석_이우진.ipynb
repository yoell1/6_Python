### 4. Age, Fare의 평균·최솟값·최댓값

```python
print("Age 평균:", df["Age"].mean().round(2))
print("Age 최댓값:", df["Age"].max().round(2))
print("Age 최솟값:", df["Age"].min().round(2))

print("Fare 평균:", df["Fare"].mean().round(2))
print("Fare 최댓값:", df["Fare"].max().round(2))
print("Fare 최솟값:", df["Fare"].min().round(2))
```

### 7. 50세 이상 탑승객 추출

```python
df_50 = df[df["Age"]>= 50] 

df_50.head()
```

### 8. 나이대(AgeGroup) 열 추가

```python
def age_group(age):
    if pd.isna(age):
        return "미확인"
    elif age < 10:
        return "아동"
    elif age < 20:
        return "10대"
    elif age < 30:
        return "20대"
    elif age < 40:
        return "30대"
    elif age < 50:
        return "40대"
    elif age < 60:
        return "50대"
    else:
        return "60대 이상"

df["AgeGroup"]= df["Age"].apply(age_group)    

df.head()
```

### 9. 성별·객실 등급별 평균 생존율

```python
df.groupby(["Sex","Pclass"])["Survived"].mean().round(2)
```

### 10. 나이대(AgeGroup)별 평균 생존율

```python
df.groupby("AgeGroup")["Survived"].mean().round(2)
```

### 11. 열별 결측치 개수와 비율 (내림차순)

```python
missing = df.isna().sum()

ratio= (missing / len(df) * 100).round(2) 

result = pd.DataFrame({
    "결측치 수": missing,
    "비율(%)": ratio
})

result.sort_values("결측치 수",ascending=False)
```

### 12. 성별 인코딩 (Gender_Encoded 열 추가)

```python
df["Gender_Encoded"] = df["Sex"].map({"male":0,"female":1})

df.head()
```

### 13. 탑승지(Embarked)별 평균 요금

```python
df.groupby("Embarked")["Fare"].mean().round(2)
```

### 14. 피벗 테이블 (Pclass × Sex, Fare 평균)

```python
df.pivot_table(index="Pclass",columns="Sex",values="Fare",aggfunc="mean").round(2)
```

### 15. 가족 수(FamilySize) 열 추가 및 요약 통계

```python
df["FamilySize"] = df["SibSp"]+ df["Parch"]

df["FamilySize"].describe().round(2)
```

### 16. 이름에서 호칭(Title) 추출

```python
df["Title"] = df["Name"].str.extract(r', ([A-Za-z]+)\.')

df["Title"].value_counts().head(5)
```

### 17. 호칭(Title)별 승객 수·평균 나이·평균 생존율

```python
df.groupby("Title").agg(
    승객수=("PassengerId","count"),
    평균나이=("Age","mean"),
    평균생존율=("Survived","mean")
).round(2)
```

### 18. 생존자·사망자 나이 분포 비교 (히스토그램)

```python
survived = df[df["Survived"]==1]["Age"]
died = df[df["Survived"]==0]["Age"]

import matplotlib.pyplot as plt

plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False

fig, ax = plt.subplots()

ax.hist(survived.dropna(), bins=20, alpha=0.5, label="생존")
ax.hist(died.dropna(), bins=20, alpha=0.5, label="사망")

ax.set_title("생존자, 사망자 나이 분포")
ax.set_xlabel("나이")
ax.set_ylabel("승객 수")
ax.legend()

fig.savefig("age_distribution.png")
plt.close()
```

### 19. Title·Pclass 그룹별 나이 중앙값으로 Age 결측치 채우기

```python
median_age =df.groupby(["Title","Pclass"])["Age"].transform("median")

df["Age"] = df["Age"].fillna(median_age)

df["Age"].isna().sum()
```

### 20. 수치형 변수 상관관계 히트맵

```python
cols = ["Survived", "Pclass", "Age", "SibSp", "Parch", "Fare"]

corr = df[cols].corr().round(2)

corr

import seaborn as sns

fig, ax = plt.subplots(figsize=(8, 6))

sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0, ax=ax)

ax.set_title("수치형 변수 상관관계 히트맵")

fig.savefig("heatmap.png")
plt.close()
```
